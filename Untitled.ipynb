```python
import pandas as pd
```

```python
from wordfreq import zipf_frequency
from pathlib import Path
```

```python
import kagglehub
# Download latest version
path = kagglehub.dataset_download("gowrishankarp/chat-slang-abbreviations-acronyms")

print("Path to dataset files:", path)
```

```python
df_json = pd.read_json(Path(path, "slang", "slang.json"), orient='index', typ = 'series')
```

```python
for i in ["i", "am", "who", "are", "job", "bf", "afk", "bmw", "hiv", "covid"]:
    x = zipf_frequency(i, 'en', wordlist='best')
    print(i, x)
```

```python
keys = df_json.keys()
```

```python
ZIPF_THRESHOLD = 3.0

KEEP = {
    "lol", "lmao", "rofl", "afk", "brb", "bro", "vro", "bruh",
    "imo", "imho", "ngl", "irl", "gg",
    "ez", "sus", "ikr", "rn", "tbh", "yw",
}

REMOVE = {
    "bmw", "amd", "intel", "nvidia", "cpu", "gpu",
    "hiv", "covid", "aids",
    "hr", "kg", "km", "cm", "mm",
}

OVERRIDE = {
    "ext": "external",
    "exts": "externals",
    "fw": "feed water",
    "fr": "for real",
    "fws": "feed waters",
    "chp": "checkpoint",
    "lo": "logistics",
    "py": "python",
    "tran": "transformer",
    "trans": "transformer",
    "wym": "what do you mean?",
    "vro": "bro",
    "ima": "iama",
    "tmp": "temporary",
    "onw": "on way",
    "pm": "p m",
    "trt": "t r t",
    "ps": "p s",
    "wt": "wait",
    "ht": "have to",
    "rm": "remove",
    "cp": "copy",
    "mv": "move",
}

removed = {}
kept = {}
```

```python
for word, value in df_json.items():
    w = word.strip().lower()
    freq = zipf_frequency(w, "en")

    if w in KEEP:
        kept[word] = value
        continue

    if w in REMOVE:
        removed[word] = value
        continue

    if freq >= ZIPF_THRESHOLD:
        removed[word] = value
    else:
        kept[word] = value
for key, value in OVERRIDE.items():
    kept[key] = value

kept = pd.Series(kept)
removed = pd.Series(removed)

print(f"Kept: {len(kept)}")
print(f"Removed: {len(removed)}")
kept.to_json(
    "slang_clean.json",
    indent=2,
    force_ascii=False,
)
```

```python
from kittentts import normalize_text
normalized = normalize_text("Dr. Rivera paid $12.50 at 3:05 p.m.")
# "Doctor Rivera paid twelve dollars and fifty cents at three oh five p m."
print(normalized)
```

```python
import re
pattern = '[a-zA-Z0-9]+'
def advance_normalize_text(text) -> str:
    normalized = normalize_text(text)
    return re.sub(pattern, lambda match: kept.get(match.group(), match.group()), normalized)

text = "this is a test2 message, I want to test the members. ima test fr but ht wt. "
advance_normalize_text(text)
```

```python
audio = model.generate(advance_normalize_text("Hellow tween"), voice="Rosie", speed=1.4)
```
